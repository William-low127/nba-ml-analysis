```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

# Set seed for reproducibility
np.random.seed(42)

# Generate base feature
x = np.linspace(0, 10, 100)
y_true = 2.5 * x + 5

# Create datasets with varying noise levels
noise_levels = [1, 5, 12, 35]
titles = ["High R² (~0.98)", "Moderate R² (~0.75)", "Low R² (~0.30)", "Near-Zero R² (~0.05)"]

fig, axes = plt.subplots(1, 4, figsize=(20, 5), sharey=True)

for i, noise in enumerate(noise_levels):
    # Add Gaussian noise
    y_noisy = y_true + np.random.normal(0, noise, size=x.shape)
    
    # Fit linear trendline
    slope, intercept = np.polyfit(x, y_noisy, 1)
    y_pred = slope * x + intercept
    
    # Calculate exact R² score
    r2 = r2_score(y_noisy, y_pred)
    
    # Plot scatter points and regression line
    axes[i].scatter(x, y_noisy, alpha=0.7, color='tab:blue', label='Data points')
    axes[i].plot(x, y_pred, color='tab:red', linewidth=2, label='Model Fit')
    axes[i].set_title(f"{titles[i]}\n$R^2 = {r2:.2f}$", fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Feature (X)')
    if i == 0:
        axes[i].set_ylabel('Target (Y)')
    axes[i].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
